## Theory
- **Machine Learning:** algorithms that learn patterns from data to make predictions on new data — no explicit rules written by hand.
- **Training loop:** feed data → model finds patterns → adjust → repeat.
- **Features (X)** = inputs. **Target (y)** = what we're predicting.
- **Three types:**
  - Supervised — labeled y is given.
  - Unsupervised — no y, just structure to find.
  - Reinforcement — an agent learns from reward/penalty feedback.

## Example 1 — Splitting a match record into Features (X) and Target (y)

```python
# One FC Lahore Lions match record
match = {
    "shots_on_target": 8,
    "possession_pct": 61,
    "corners": 6,
    "result": "Win"   # this is what we want to predict
}

# Features = everything except the thing we're predicting
X = {k: v for k, v in match.items() if k != "result"}
# Target = the thing we're predicting
y = match["result"]

print("Features (X):", X)
print("Target (y):", y)
```

## Example 2 — A tiny 'training loop' (intuition only, no real model math yet)

```python
# Simplified training loop — no gradient descent yet, just the idea of
# 'look at past data, adjust the guess'
past_results = ["win", "loss", "win", "win", "draw"]

def naive_model(history):
    # a 'model' here is just: guess the most frequent past result
    return max(set(history), key=history.count)

prediction = naive_model(past_results)
print("Model's current guess for the next result:", prediction)
```
