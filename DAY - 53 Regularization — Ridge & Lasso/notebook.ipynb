# Day 53 — Regularization: Ridge & Lasso

## Objective- Understand why plain Linear Regression overfits with correlated/too many features- Learn how Ridge (L2) and Lasso (L1) penalties fix this- Compare coefficients and test R² across LinearRegression, Ridge, and Lasso

## Imports

```python
import numpy as npimport pandas as pdfrom sklearn.linear_model import LinearRegression, Ridge, Lassofrom sklearn.model_selection import train_test_splitfrom sklearn.preprocessing import StandardScalerfrom sklearn.datasets import make_regressionnp.random.seed(42)  # keep results reproducible
```

## Theory**Regularization** adds a penalty term to the cost function so the model can't assign huge weights to any single feature.- **Ridge (L2):** penalty = λ × Σ(w²) — shrinks all weights, keeps every feature- **Lasso (L1):** penalty = λ × Σ|w| — can shrink some weights to exactly 0 = automatic feature selectionBoth need scaled features first, since the penalty is sensitive to the scale of the weights.

## Example 1 — Baseline overfitting with correlated features

```python
# Build a dataset where some features are correlated (redundant)X, y = make_regression(n_samples=100, n_features=5, n_informative=3,                        noise=15, random_state=42)X_train, X_test, y_train, y_test = train_test_split(    X, y, test_size=0.3, random_state=42)scaler = StandardScaler().fit(X_train)X_train_s = scaler.transform(X_train)X_test_s = scaler.transform(X_test)lr = LinearRegression().fit(X_train_s, y_train)print('Train R2:', round(lr.score(X_train_s, y_train), 3))print('Test  R2:', round(lr.score(X_test_s, y_test), 3))
```

## Example 2 — Ridge vs Lasso coefficients

```python
ridge = Ridge(alpha=1.0).fit(X_train_s, y_train)lasso = Lasso(alpha=0.5).fit(X_train_s, y_train)coef_table = pd.DataFrame({    'LinearRegression': lr.coef_.round(2),    'Ridge': ridge.coef_.round(2),    'Lasso': lasso.coef_.round(2),})coef_table
```

## Practice ExerciseFit Ridge and Lasso with **three different alpha values each** (e.g. 0.1, 1, 10) and print the R² on the test set for every run. Which alpha gives the best test R² for each model?

```python
# Your turn — fill this infor a in [0.1, 1, 10]:    r = Ridge(alpha=a).fit(X_train_s, y_train)    l = Lasso(alpha=a).fit(X_train_s, y_train)    print(f'alpha={a:>4} | Ridge R2: {r.score(X_test_s, y_test):.3f} | Lasso R2: {l.score(X_test_s, y_test):.3f}')
```

## Mini ChallengeCount how many coefficients Lasso zeroes out at `alpha=1.0` vs `alpha=5.0`. What does that tell you about using Lasso for feature selection as alpha increases?

```python
for a in [1.0, 5.0]:    l = Lasso(alpha=a).fit(X_train_s, y_train)    zeroed = (l.coef_ == 0).sum()    print(f'alpha={a} -> {zeroed} of {len(l.coef_)} coefficients zeroed out')
```

## Summary- Regularization = penalty on the cost function to fight overfitting- Ridge (L2) shrinks all weights, keeps every feature- Lasso (L1) can zero out weights — doubles as feature selection- Always scale features first, and compare against a plain LinearRegression baseline- Alpha/λ controls regularization strength — too high underfits, too low barely helps
