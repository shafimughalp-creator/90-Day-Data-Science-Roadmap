# OUTER JOINs

Day 40 — SQL Phase, FC Lahore Lions SQL Lab

## Objective

Learn how LEFT JOIN, RIGHT JOIN, and FULL OUTER JOIN keep unmatched rows instead of dropping them, and use the anti-join pattern (`OUTER JOIN` + `IS NULL`) to find missing relationships — e.g. a player with no team, or a team with no players.

## Imports

```
import sqlite3
import pandas as pd

# connect to today's practice database
conn = sqlite3.connect('outer_joins.db')
print('Connected. SQLite version:', sqlite3.sqlite_version)
```

## Theory

- **INNER JOIN** keeps only rows that match on both sides.
- **LEFT JOIN** keeps every row from the left (first) table, filling unmatched right-side columns with `NULL`.
- **RIGHT JOIN** keeps every row from the right (second) table, filling unmatched left-side columns with `NULL`.
- **FULL OUTER JOIN** keeps every row from both tables.
- **Anti-join** = OUTER JOIN + `WHERE <col> IS NULL` → isolates only the unmatched rows.
- `= NULL` never matches — always use `IS NULL` / `IS NOT NULL`.

## Example 1 — LEFT JOIN: every player, team or no team

```
query = """
SELECT p.player_name, p.goals, t.team_name
FROM players p
LEFT JOIN teams t ON p.team_id = t.team_id;
"""
pd.read_sql_query(query, conn)
```

## Example 2 — Anti-join: players with no team

```
query = """
SELECT p.player_name, p.goals
FROM players p
LEFT JOIN teams t ON p.team_id = t.team_id
WHERE t.team_id IS NULL;
"""
pd.read_sql_query(query, conn)
```

## Practice Exercise

Write a RIGHT JOIN that lists every team, along with any players it has (or NULL if it has none).

```
# Your turn: RIGHT JOIN teams to players
query = """
SELECT t.team_name, p.player_name
FROM players p
RIGHT JOIN teams t ON p.team_id = t.team_id;
"""
pd.read_sql_query(query, conn)
```

## Mini Challenge

Find the team(s) with zero players, using the anti-join pattern on a RIGHT JOIN. Then use `FULL OUTER JOIN` to get *both* the unmatched player and the unmatched team in a single query.

```
# Teams with zero players
query = """
SELECT t.team_name
FROM players p
RIGHT JOIN teams t ON p.team_id = t.team_id
WHERE p.player_id IS NULL;
"""
print('Teams with no players:')
print(pd.read_sql_query(query, conn))

# Everyone, matched or not
query_full = """
SELECT p.player_name, t.team_name
FROM players p
FULL OUTER JOIN teams t ON p.team_id = t.team_id;
"""
print('\nFULL OUTER JOIN result:')
print(pd.read_sql_query(query_full, conn))
```

## Summary

- OUTER JOINs keep unmatched rows; INNER JOIN silently drops them.
- LEFT/RIGHT decide *which* table's unmatched rows survive; FULL keeps both.
- The anti-join pattern (OUTER JOIN + `IS NULL`) is the standard way to find missing relationships — one of the most common SQL interview questions.
