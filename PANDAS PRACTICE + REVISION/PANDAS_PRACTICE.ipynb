⚽ Day 20: Pandas — Series & DataFrame Fundamentals

```python
import pandas as pd

# Creating a DataFrame from a Python dictionary
data = {
    "Player": ["Saka", "Salah", "Rodrygo", "Palmer"],  # Name column
    "Position": ["RW", "RW", "RW", "RW"],  # Right Winger position
    "Goals": [16, 19, 10, 22],  # Total goals scored
    "Assists": [9, 10, 5, 11],  # Total assists provided
}

df = pd.DataFrame(data)  # Convert dictionary to DataFrame
print(df)
```

```python
print(df.columns)
```

☕ Day 21: Pandas — Cleaning: Missing Values & Duplicates

```python
import numpy as np
import pandas as pd

coffee_data = {
    "Order_ID": [101, 102, 102, 103, 104, 105],  # Notice 102 is duplicated
    "Item": ["Espresso", "Latte", "Latte", "Cappuccino", np.nan, "Mocha"],  # NaN item
    "Price": [3.0, 4.5, 4.5, np.nan, 3.5, 4.0],  # NaN price
    "Quantity": [1, 2, 2, 1, 3, 1],
}

df_coffee = pd.DataFrame(coffee_data)
```

```python
# 1. Detecting missing values
missing_counts = df_coffee.isnull().sum()
print(missing_counts)
```

```python
# 2. Dropping rows with missing values
df_dropped = df_coffee.dropna(subset=['Item','Price'])
print(df_dropped)
```

```python
# 3. Filling missing values (Imputation)
avg_price = df_coffee['Price'].mean()
print(avg_price)
```

```python
df_coffee['Price'] = df_coffee['Price'].fillna(avg_price)
print(df_coffee)
```

```python
# 4. Detecting and dropping duplicates
dups = df_coffee.duplicated()
print(dups)
```

```python
# Anatomy: df.drop_duplicates(subset=[cols], keep='first', inplace=False)
df_clean = df_coffee.drop_duplicates(keep="first")
print(df_clean)
```

🎓 Day 22: Pandas — Data Types & Type Conversion

```python
student_data = {
    "Student_ID": ["S1", "S2", "S3", "S4"],
    "Score": [
        "85",
        "90",
        "Missing",
        "78",
    ],  # Numbers stored as strings, one text error
    "Exam_Date": ["2026-03-01", "2026-03-02", "2026-03-02", "2026-03-03"],  # Dates as strings
    "Passed": ["1", "1", "0", "1"],  # Integers stored as strings
}

df_students = pd.DataFrame(student_data)
print(df_students.dtypes)  # Everything is currently stored as 'object'!
```

```python
# 1. Converting text to numeric values
df_students['Score'] = pd.to_numeric(df_students['Score'],errors="coerce")
print(df_students.dtypes)
```

```python
# 2. Converting strings to standard datetime objects
df_students['Exam_Date'] = pd.to_datetime(df_students['Exam_Date'])
df_students.dtypes
```

```python
# 3. Casting data types using .astype()
df_students['Passed'] = df_students['Passed'].astype(int)
df_students.dtypes
```

```python
df_students['Student_ID'] = df_students['Student_ID'].astype("category")
df_students.dtypes
```

🛒 Day 23: Pandas — GroupBy & Aggregation

```python
sales_data = {
    "Category": ["Electronics", "Clothing", "Electronics", "Clothing", "Electronics", "Clothing"],
    "Region": ["North", "North", "South", "South", "North", "South"],
    "Price": [500, 50, 800, 60, 300, 40],
    "Quantity": [2, 5, 1, 10, 3, 8],
}

df_sales = pd.DataFrame(sales_data)
# Create a total revenue column for our calculations
df_sales["Revenue"] = df_sales["Price"] * df_sales["Quantity"]
df_sales
```

```python
# 1. Single aggregation on a single grouped column
cat_revenue = df_sales.groupby("Category")['Revenue'].sum()
print(cat_revenue)
```

```python
# 2. Multi-column grouping with multiple aggregations
multi_group = df_sales.groupby(["Category","Region"]).agg({
    "Revenue" : ["sum","mean"] ,
    "Quantity" : "sum"
})
print(multi_group)
```

```python
# 3. Named aggregations (The clean, modern industry standard)
top_categories = df_sales.groupby("Category").agg(
    Total_Revenue=("Revenue","sum") ,
    Avg_price=("Price","mean") ,
    Total_Units=("Quantity","sum")
).sort_values("Total_Revenue",ascending=False)
print(top_categories)
```

🔗 Day 24: Pandas — Merge, Join & Concat

```python
# Table 1: Customers
customers = pd.DataFrame(
    {"Customer_ID": [1, 2, 3], "Name": ["Ali", "Bilal", "Hamza"]}
)

# Table 2: Orders (Notice Customer_ID 4 does not exist in Customers table!)
orders = pd.DataFrame(
    {
        "Order_ID": [101, 102, 103, 104],
        "Customer_ID": [1, 2, 1, 4],
        "Product_ID": ["P1", "P2", "P1", "P3"],
    }
)

# Table 3: Products
products = pd.DataFrame(
    {"Product_ID": ["P1", "P2", "P3"], "Price": [1000, 500, 200]}
)
```

```python
# 1. Inner Join: Keeps ONLY matching rows in both tables
inner_merge = pd.merge(customers , orders , on="Customer_ID" , how="inner")
print(inner_merge)
```

```python
# 2. Left Join: Keeps ALL rows from the left table, fills NaN for missing matches
left_merge = pd.merge(customers , orders , on="Customer_ID" , how="left")
print(left_merge)
```

```python
# 3. Three-Table Join Chain
full_pipeline = customers.merge(orders, on="Customer_ID", how="inner").merge(
    products, on="Product_ID", how="left"
)
print(full_pipeline)
```

```python
# 4. Vertical Concatenation: Stacking DataFrames on top of each other
new_customers = pd.DataFrame({
    "Customer_ID" : [4,5] ,
    "Name" : ["Shafi","Shafay"]
})
all_customers = pd.concat([customers , new_customers],ignore_index=True)
print(all_customers)
```

🔤 Day 25: Pandas — Apply, Map, Lambda & String Operations

```python
emp_data = {
    "Emp_ID": [101, 102, 103, 104],
    "Name": [
        "  JOHN DOE  ",
        "sarah connor",
        "Bruce Wayne",
        "clark KENT  ",
    ],  # Messy spacing & case
    "Dept_Code": ["ENG", "HR", "ENG", "MKT"],
    "Salary": [80000, 65000, 95000, 70000],
}

df_emp = pd.DataFrame(emp_data)
```

```python
# 1. Core String Operations using .str accessor
df_emp['Name_Clean'] = df_emp['Name'].str.strip().str.title()
print(df_emp)
```

```python
df_emp['Is_engineering'] = df_emp['Dept_Code'].str.contains('ENG')
print(df_emp)
```

```python
# 2. Value replacement using .map()
dept_names = {"ENG":"Engineering","HR":"Human Resources","MKT":"Marketing"}
df_emp['Department'] = df_emp['Dept_Code'].map(dept_names)
print(df_emp)
```

```python
# 3. Custom logic across a single column using .apply() and lambda
df_emp['Bonus'] = df_emp['Salary'].apply(
    lambda x : x * 0.10 if x < 75000 else x * 0.05
)
print(df_emp)
```

```python
# 4. Row-level calculations across multiple columns (axis=1)
df_emp["Total_Compensation"] = df_emp.apply(
    lambda row: row["Salary"] + row["Bonus"], axis=1
)

print(df_emp[["Name_Clean", "Department", "Total_Compensation"]])
```

📊 Day 26: Pandas — Sorting, Ranking, Rolling & Pivot Tables

```python
revenue_data = {
    "Date": pd.to_datetime(
        [
            "2026-07-01",
            "2026-07-02",
            "2026-07-03",
            "2026-07-04",
            "2026-07-01",
            "2026-07-02",
            "2026-07-03",
            "2026-07-04",
        ]
    ),
    "Store": [
        "Lahore",
        "Lahore",
        "Lahore",
        "Lahore",
        "Karachi",
        "Karachi",
        "Karachi",
        "Karachi",
    ],
    "Revenue": [100, 150, 130, 200, 300, 280, 350, 400],
}

df_rev = pd.DataFrame(revenue_data)
```

```python
# 1. Sorting by columns
df_sorted = df_rev.sort_values(["Store","Date"],ascending=[True,True])
print(df_sorted)
```

```python
# 2. Ranking values within groups
df_rev["Revenue_Rank"] = df_rev.groupby("Store")["Revenue"].rank(
    method="dense" , ascending=False
)
print(df_sorted)
```

```python
# 3. Rolling Window (Moving Average for Feature Engineering)
df_rev["2_Day_MA"] = (
    df_rev.groupby("Store")["Revenue"]
    .rolling(window=2)
    .mean()
    .reset_index(level=0, drop=True)
)
print(df_rev["2_Day_MA"])
```

```python
pivot_report = pd.pivot_table(
    df_rev,
    values="Revenue",  # The numbers to aggregate
    index="Date",  # Rows of the pivot table
    columns="Store",  # Columns of the pivot table
    aggfunc="sum",  # Aggregation function
)

print(pivot_report)
```
