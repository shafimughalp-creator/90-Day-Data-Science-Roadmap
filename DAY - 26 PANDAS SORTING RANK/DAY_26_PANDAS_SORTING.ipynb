```python
import pandas as pd

# Orbit Mart — 20 daily orders across regions, countries, and product categories
data = {
    'date':     pd.date_range(start='2024-01-01', periods=20, freq='D'),  # 20 sequential days
    'region':   ['South Asia','South Asia','South Asia','South Asia','Europe','Europe',
                 'Europe','Europe','North America','North America','North America','North America',
                 'South Asia','South Asia','Europe','Europe','North America','North America',
                 'South Asia','Europe'],
    'country':  ['Pakistan','Pakistan','India','India','UK','UK','Germany','Germany',
                 'USA','USA','Canada','Canada','Pakistan','India','UK','Germany',
                 'USA','Canada','Pakistan','UK'],
    'category': ['Electronics','Fashion','Electronics','Grocery','Fashion','Electronics',
                 'Grocery','Fashion','Electronics','Grocery','Fashion','Electronics',
                 'Grocery','Fashion','Grocery','Electronics','Fashion','Grocery',
                 'Electronics','Fashion'],
    'sales':    [15000, 8000, 21000, 6000, 17000, 25000, 9000, 14000, 32000, 11000,
                 13000, 19000, 7000, 10000, 8000, 23000, 16000, 5000, 18000, 20000],
    'clv':      [52000, 31000, 61000, 18000, 72000, 88000, 27000, 49000, 110000, 39000,
                 45000, 67000, 22000, 35000, 29000, 79000, 58000, 17000, 64000, 73000],
}
df = pd.DataFrame(data)
print(df.head())
```

```python
top_orders = df.sort_values('sales',ascending=False).head(10)
print(top_orders[['country','category','sales']])
```

Now, rank() — giving every row a position number:

```python
df['sales_rank'] = df['sales'].rank(method='dense' , ascending=False)
print(df[['country','sales','sales_rank']].sort_values('sales_rank').head(6))
```

```python
# Rank only Electronics orders by sales, using a boolean mask + rank together
electronics_mask = df['category'].str.contains('Electronics')   # Day 25: .str method
df.loc[electronics_mask, 'elec_rank'] = df.loc[electronics_mask, 'sales'].rank(ascending=False)
print(df.loc[electronics_mask, ['country', 'sales', 'elec_rank']])
```

Sort df by clv (customer lifetime value), highest first, and show the top 5 rows (country, clv only).

```python
customer_lifetime_value = df.sort_values('clv',ascending=False).head(5)
print(customer_lifetime_value[['country','sales','clv']])
```

Add a new column clv_rank that ranks every order by clv (1 = highest), using method='min' instead of 'dense' this time.

```python
# Fix 1 — include clv_rank in the column list you select
print(df[['country', 'sales', 'clv', 'clv_rank']].sort_values('clv_rank'))

# Fix 2 — sort first (on the full df), select columns after
print(df.sort_values('clv_rank')[['country', 'sales', 'clv', 'clv_rank']])
```

```python
df['rolling_mean_sales'] = df['sales'].rolling(window=7).mean()   # avg of current + previous 6 rows
print(df[['date', 'sales', 'rolling_mean_sales']])
```

```python
above_avg_mask = df['sales'] > df['rolling_mean_sales']   # boolean mask (NumPy/Pandas comparison)
df['beat_rolling_avg'] = above_avg_mask                    # store it as a flag column
print(df.loc[above_avg_mask, ['date', 'sales', 'rolling_mean_sales']])   # only the days that beat their own trend
```

```python
df['rolling_3day_max'] = df['sales'].rolling(window=3, min_periods=1).max()
df  # max of last 3 rows
```

Create rolling_3day_mean — a 3-row rolling average of clv, using min_periods=1 so there are no NaNs at the start.

```python
df['rolling_3day_mean'] = df['clv'].rolling(window=3 , min_periods=1).mean()
print(df[['sales','clv','rolling_3day_mean']])
```

Create a boolean column clv_above_trend that's True wherever clv is greater than rolling_3day_mean.

```python
clv_above_trend = df['clv'] > df['rolling_3day_mean']
df['is_clv_above_trend'] = clv_above_trend
print(df.loc[clv_above_trend,['sales','clv','rolling_3day_mean']])
```

```python
region_category_sales = pd.pivot_table(
    df,
    values='sales',       # what to aggregate
    index='region',       # becomes the new rows
    columns='category',   # becomes the new columns
    aggfunc='sum'          # how to combine multiple matching rows
)
print(region_category_sales)
```

```python
region_category_sales_sorted = region_category_sales.sort_values('Electronics', ascending=False)
print(region_category_sales_sorted)
```

```python
country_category_clv = pd.pivot_table(
    df,
    values='clv',
    index=['region', 'country'],   # two levels: region, then country nested inside it
    columns='category',
    aggfunc='mean'
)
print(country_category_clv)
```

Build a pivot table: average clv by country (rows) and category (columns) — using aggfunc='mean'.

```python
avg_clv_country = pd.pivot_table(
    df,
    values='clv' ,
    index='country' ,
    columns='category' ,
    aggfunc='mean'
)
print(avg_clv_country)
```
