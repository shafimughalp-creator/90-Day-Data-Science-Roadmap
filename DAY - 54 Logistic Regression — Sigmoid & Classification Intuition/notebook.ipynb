# Day 54 — Logistic Regression: Sigmoid & Classification Intuition

## Objective- Understand why plain Linear Regression fails for classification- Learn the sigmoid function and plot its S-curve- Understand decision boundaries and why log loss (not MSE) is used for classification

## Imports

```python
import numpy as npimport matplotlib.pyplot as pltfrom sklearn.linear_model import LinearRegression, LogisticRegressionfrom sklearn.model_selection import train_test_splitfrom sklearn.metrics import log_loss, mean_squared_errorfrom sklearn.datasets import make_classificationnp.random.seed(42)  # keep results reproducible
```

## Theory**Why linear regression fails for classification:** its output is unbounded, so it can predict values below 0 or above 1 for a 0/1 label.**Sigmoid function:** σ(z) = 1 / (1 + e⁻ᶻ) — squashes any real number into [0, 1], so it can be read as a probability.**Decision boundary:** predict class 1 if σ(z) > 0.5, else class 0. The 0.5 threshold can be moved.**Log loss vs MSE:** log loss penalizes confident wrong predictions much more heavily than MSE does — which is why it's the standard loss for classification.

## Example 1 — Linear regression breaks on 0/1 labels

```python
X, y = make_classification(n_samples=100, n_features=1, n_informative=1,                            n_redundant=0, n_clusters_per_class=1, random_state=42)lr = LinearRegression().fit(X, y)preds = lr.predict(X)print('Min prediction:', preds.min().round(2))print('Max prediction:', preds.max().round(2))print('These are not valid probabilities!')
```

## Example 2 — Sigmoid squashes scores into [0, 1] and plots the S-curve

```python
def sigmoid(z):    return 1 / (1 + np.exp(-z))z = np.linspace(-10, 10, 200)plt.plot(z, sigmoid(z))plt.axhline(0.5, ls='--', color='gray')plt.axvline(0, ls='--', color='gray')plt.title('Sigmoid function: maps any z to [0, 1]')plt.xlabel('z')plt.ylabel('sigma(z)')plt.show()
```

## Mini ChallengeCompute log loss AND mean squared error on the same set of predicted probabilities vs true labels. Which one changes more when a prediction is confidently wrong?

```python
y_true = np.array([1, 0, 1, 0])y_pred_proba = np.array([0.99, 0.95, 0.51, 0.49])  # 2nd pred is confidently WRONGprint('Log loss:', log_loss(y_true, y_pred_proba))print('MSE:     ', mean_squared_error(y_true, y_pred_proba))
```

## Summary- Linear regression's unbounded output doesn't work for 0/1 classification labels- Sigmoid σ(z) = 1/(1+e⁻ᶻ) squashes any score into a valid [0, 1] probability- A decision boundary (default 0.5) turns that probability into a predicted class- Log loss, not MSE, is used to train/evaluate classifiers because it punishes confident mistakes much harder- These sigmoid + log loss ideas reappear directly inside neural networks
