```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Dataset for weekend matches
data = {
    'day': ['Saturday', 'Saturday', 'Saturday', 'Saturday', 'Saturday', 'Saturday',
            'Sunday', 'Sunday', 'Sunday', 'Sunday', 'Sunday', 'Sunday'],
    'shots_on_target': [7, 3, 9, 2, 5, 8, 4, 1, 6, 10, 3, 5],
    'result': ['Win', 'Loss', 'Win', 'Loss', 'Draw', 'Win',
               'Draw', 'Loss', 'Win', 'Win', 'Loss', 'Draw']
}

df_matches = pd.DataFrame(data)

sns.catplot(
    data=df_matches ,
    x='day' ,
    y='shots_on_target' ,
    col='result' ,
    kind='box' ,
)

plt.show()
```

```python
g = sns.FacetGrid(df_matches, col='result')   # Creates 3 blank plots side-by-side for Win, Loss, Draw
g.map(sns.histplot, 'shots_on_target')        # Fills each plot with a histogram of shots
plt.savefig('match_shots_grid.png', dpi=200, bbox_inches='tight')  # Saves high-quality PNG
```

```python
# 1. Create a Matplotlib grid: 1 row, 2 columns
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# 2. Draw a Seaborn boxplot and force it into the LEFT box (axes[0])
sns.boxplot(data=df_matches, x='day', y='goals', ax=axes[0])

# 3. Draw a Seaborn histplot and force it into the RIGHT box (axes[1])
sns.histplot(data=df_matches, x='shots_on_target', ax=axes[1])

plt.show()
```

```python
# 1. Create the 1x2 grid
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 2. Filter data for Wins and plot on the left
wins = df_matches[df_matches['result'] == 'Win']              # Boolean mask for Wins
sns.histplot(data=wins, x='shots_on_target', ax=axes[0])      # Placed in left box
axes[0].set_title('Shots on Target (Wins)')                   # Matplotlib title

# 3. Filter data for Losses and plot on the right
losses = df_matches[df_matches['result'] == 'Loss']           # Boolean mask for Losses
sns.histplot(data=losses, x='shots_on_target', ax=axes[1])    # Placed in right box
axes[1].set_title('Shots on Target (Losses)')                 # Matplotlib title

# 4. Save professional output
plt.savefig('wins_vs_losses.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Initialize the grid (slightly wider to fit both charts comfortably)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 2. Histogram on the left
sns.histplot(
    data=df_matches,
    x='shots_on_target',
    ax=axes[0]
)

# 3. Boxplot on the right
sns.boxplot(
    data=df_matches,
    x='day',
    y='goals',
    ax=axes[1]
)

# 4. Save the figure before showing it!
plt.savefig('winger_dashboard.png', dpi=200, bbox_inches='tight')

# 5. Display
plt.show()
```
