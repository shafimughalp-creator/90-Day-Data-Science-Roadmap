```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data: Match weeks and winger performance metrics
matches = np.array([1, 2, 3, 4, 5])                  # X-axis: Match numbers
shots = np.array([3, 5, 2, 6, 4])                    # Y-axis 1: Shots on target
passes = np.array([25, 30, 18, 35, 28])              # Y-axis 2: Completed passes

# Create a 2x2 subplot grid on a single canvas
fig , axes = plt.subplots(nrows=2 , ncols=2 , figsize=(10,8))
# Plot 1: Top-Left (Row 0, Column 0)
axes[0,0].plot(matches , shots , color="red" , marker="o")
axes[0,0].set_title("Shots On Target")

# Plot 2: Top-Right (Row 0, Column 1)
axes[0,1].bar(matches , passes , color="blue")
axes[0,1].set_title("Passes Completion")

plt.tight_layout()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Creating the Winger Match Analytics DataFrame
data = {
    'match_week': [1, 2, 3, 4],
    'sprint_speed_kmh': [28, 31, 29, 33],
    'tackles_won': [2, 4, 1, 5]
}

df = pd.DataFrame(data)

# creating 1 v 2 grid (1 row, 2 columns -> 1D array!)
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

# left plot (using single index axes[0] and DataFrame df)
axes[0].plot(df["match_week"], df["sprint_speed_kmh"], color="yellow", marker="o")
axes[0].set_title("Sprint Speed (km/h)")

# right plot (using single index axes[1] and DataFrame df)
axes[1].bar(df["match_week"], df["tackles_won"], color="black")
axes[1].set_title("Tackles Won")

# apply tight layout
plt.tight_layout()

plt.show()  # Displays the chart!
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Apply a professional stylesheet BEFORE creating the figure
plt.style.use('seaborn-v0_8-darkgrid')                         # Adds dark grid background with clean typography

df = pd.DataFrame({
    'match_week': [1, 2, 3, 4],
    'left_wing_shots': [3, 5, 2, 6],
    'right_wing_shots': [1, 2, 4, 3] ,
    'sprint_speed_kmh': [28, 31, 29, 33] ,
    'tackles_won': [2, 4, 1, 5]
})

# 2. Create figure and single axes
fig, ax = plt.subplots(figsize=(8, 5))

# 3. Plot two lines on the same axes and assign a 'label' for the legend
ax.plot(df['match_week'], df['left_wing_shots'], color='red', marker='o', label='Left Wing')   # Line 1
ax.plot(df['match_week'], df['right_wing_shots'], color='blue', marker='s', label='Right Wing') # Line 2

# 4. Add clear Axis Labels and Title
ax.set_title("Winger Shot Comparison by Flank", fontsize=14, fontweight='bold')
ax.set_xlabel("Fixture Week", fontsize=12)                     # Names the X-axis
ax.set_ylabel("Shots on Target", fontsize=12)                  # Names the Y-axis

# 5. Display the legend (uses the 'label' arguments from above)
ax.legend(loc='upper left')                                    # Places box in top-left corner

plt.tight_layout()
```

```python
# Combining: Stylesheet + 1x2 Subplots + Pandas Filtering + Labels/Legends
plt.style.use('seaborn-v0_8-whitegrid')                        # Clean white grid style

# Filter DataFrame using Boolean Masking (Only weeks where sprint speed > 28)
fast_weeks = df[df['sprint_speed_kmh'] > 28]

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 4))

# Left Plot: Filtered Sprint Speed
axes[0].plot(fast_weeks['match_week'], fast_weeks['sprint_speed_kmh'], color='green', marker='^', label='Peak Speed')
axes[0].set_title("High Intensity Weeks (>28 km/h)")
axes[0].set_xlabel("Match Week")
axes[0].set_ylabel("Speed (km/h)")
axes[0].legend(loc='lower right')

# Right Plot: Full Tackles Bar Chart
axes[1].bar(df['match_week'], df['tackles_won'], color='purple', label='Tackles')
axes[1].set_title("Defensive Work Rate")
axes[1].set_xlabel("Match Week")
axes[1].set_ylabel("Count")
axes[1].legend(loc='upper left')

plt.tight_layout()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

plt.style.use('seaborn-v0_8-darkgrid')

df = pd.DataFrame({
    'match_week': [1, 2, 3, 4],
    'sprint_speed_kmh': [28, 31, 29, 33]
})

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(df['match_week'], df['sprint_speed_kmh'], color='yellow', marker='o')

# 1. Add text and an arrow pointing to the highest speed
ax.annotate(
    "Peak Speed!",                                       # The text to display
    xy=(4, 33),                                          # The (x, y) coordinates of the data point to point to
    xytext=(2.5, 32),                                    # The (x, y) coordinates where the text box will sit
    arrowprops=dict(facecolor='white', arrowstyle='->')  # Dictionary defining the arrow's look
)

# 2. Save the figure to your computer as a high-quality PNG
plt.savefig('winger_speed_analysis.png', dpi=300, bbox_inches='tight')

# 3. Show the plot (Always call show AFTER savefig)
plt.show()
```

📋 Project Requirements: The Tournament Dashboard
Write a single script to generate a 2x2 dashboard grid (2 rows, 2 columns):

Setup: Apply the 'seaborn-v0_8-darkgrid' style. Create a 2x2 subplot grid with figsize=(12, 8).

Top-Left (axes[0, 0]): Scoring Record

Plot goals_for vs match as a line plot (color 'maroon', label 'Scored').

On the same axes, plot goals_against vs match as a line plot (color 'grey', label 'Conceded').

Add a legend in the 'upper right'.

Set the title to "Scoring Record".

Top-Right (axes[0, 1]): Possession

Plot possession_pct vs match as a bar chart (color 'black').

Set the title to "Match Possession (%)".

Bottom-Left (axes[1, 0]): High-Scoring Matches

Filter step: Create a new DataFrame containing only matches where goals_for >= 2.

Plot a bar chart of goals_for vs match using this filtered data (color 'yellow').

Set the title to "Matches with 2+ Goals".

Bottom-Right (axes[1, 1]): The Peak

Plot goals_for vs match as a line plot (color 'maroon').

Use ax.annotate() to point an arrow at the peak scoring match (Group C, 3 goals). The coordinate to point to is xy=(2, 3). Place your text "Tournament Peak!" at xytext=(0.5, 2.5) with a white arrow.

Set the title to "Scoring Peak".

Export: Apply plt.tight_layout(), then save the dashboard as "tournament_dashboard.png" with dpi=300 and bbox_inches='tight'.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Tournament Dataset
qatar_cup = pd.DataFrame({
    'match': ['Group A', 'Group B', 'Group C', 'Ro16', 'Quarter', 'Semi'],
    'goals_for': [2, 1, 3, 2, 1, 0],
    'goals_against': [0, 1, 1, 0, 2, 2],
    'possession_pct': [55, 48, 60, 52, 45, 40]
})
# Apply the 'seaborn-v0_8-darkgrid' style. Create a 2x2 subplot grid with figsize=(12, 8).
plt.style.use("seaborn-v0_8-darkgrid")
fig , axes = plt.subplots(nrows=2 , ncols=2 , figsize=(12,8))
# Top-Left (axes[0, 0]): Scoring Record
axes[0,0].plot(qatar_cup['match'], qatar_cup['goals_for'] , color='maroon' , label="Scored")
axes[0,0].plot(qatar_cup['match'] , qatar_cup['goals_against'] , color='grey' , label='Conceded')
axes[0,0].legend(loc="upper right")

# Top-Right (axes[0, 1]): Possession
axes[0,1].bar(qatar_cup["match"],qatar_cup["possession_pct"] , color='black')
axes[0,1].set_title("Match Possession (%)")

# Bottom-Left (axes[1, 0]): High-Scoring Matches
Goals_check = qatar_cup[qatar_cup['goals_for'] >= 2]
axes[1,0].bar(Goals_check["match"] , Goals_check["goals_for"] , color="yellow")
axes[1,0].set_title("Matches with 2+ Goals")

# Bottom-Right (axes[1, 1]): The Peak
axes[1,1].plot(qatar_cup["match"] , qatar_cup["goals_for"] , color="maroon")
axes[1,1].annotate(
    "Tournament Peak" ,
    xy=(2,3) ,
    xytext=(0.5,2.5) ,
    arrowprops=dict(facecolor='white' , arrowstyle="->")
)
axes[1,1].set_title("Scoring Peak")

plt.tight_layout()

plt.savefig("tournament_dashboard.png", dpi=300 , bbox_inches='tight')
```
