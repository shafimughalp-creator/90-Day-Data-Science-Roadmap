```python
# 1. Create a dataset of football player match ratings
df_ratings = pd.DataFrame({
    'position': ['Striker', 'Striker', 'Striker', 'Defender', 'Defender', 'Defender', 'Defender'],
    'rating': [8.5, 9.0, 4.0, 7.2, 7.5, 7.1, 7.8]  # Notice the 4.0 outlier for the Striker!
})

# 2. Create the boxplot comparing the two categories
sns.boxplot(data=df_ratings, x='position', y='rating') # x is the category, y is the number

# 3. Display the plot
plt.show()
```

```python
# 1. Filter the DataFrame first (Pandas boolean mask)
# We only want to plot Defenders
defenders_only = df_ratings[df_ratings['position'] == 'Defender']

# 2. Feed the filtered DataFrame to Seaborn
sns.boxplot(data=defenders_only, x='position', y='rating')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Create a dataset of numeric house features
df_houses = pd.DataFrame({
    'size_sqft': [1500, 2000, 2500, 1200, 3000],
    'bedrooms': [3, 4, 4, 2, 5],
    'price_k': [300, 450, 500, 200, 650]
})

# 2. Calculate the correlation matrix FIRST (Pandas)
corr_matrix = df_houses.corr()

# 3. Feed the correlation matrix into Seaborn's heatmap
sns.heatmap(data=corr_matrix, annot=True, cmap='coolwarm')

# 4. Display the plot
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. The Dataset
coffee_data = pd.DataFrame({
    'store_type': ['Drive-Thru', 'Dine-In', 'Drive-Thru', 'Dine-In', 'Drive-Thru'],
    'temperature_c': [15, 22, 18, 25, 10],
    'iced_coffee_sales': [100, 200, 150, 250, 50],
    'hot_coffee_sales': [300, 150, 250, 100, 400]
})

# 2. Distribution Check (Is iced coffee sales skewed?)
sns.histplot(data=coffee_data, x='iced_coffee_sales', kde=True)
plt.show()

# 3. Outlier & Range Check (Do Drive-Thrus sell more hot coffee?)
sns.boxplot(data=coffee_data, x='store_type', y='hot_coffee_sales')
plt.show()

# 4. Correlation Check (Does temperature affect sales?)
# Drop the text column before running .corr()
numeric_coffee = coffee_data[['temperature_c', 'iced_coffee_sales', 'hot_coffee_sales']]
sns.heatmap(data=numeric_coffee.corr(), annot=True, cmap='coolwarm')
plt.show()

# 5. The Ultimate Overview (Everything at once, colored by store type)
sns.pairplot(data=coffee_data, hue='store_type')
plt.show()
```
