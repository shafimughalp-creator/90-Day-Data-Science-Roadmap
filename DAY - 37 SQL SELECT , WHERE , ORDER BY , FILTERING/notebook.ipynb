# Day 37: SQL Filtering, Sorting & NULLs

Beginner-friendly hands-on practice — WHERE, AND/OR, ORDER BY, DISTINCT, IS NULL.

## Objective
- Filter rows with `WHERE`
- Combine conditions with `AND` / `OR`
- Sort results with `ORDER BY`
- Get unique values with `DISTINCT`
- Handle missing data with `IS NULL` / `IS NOT NULL`

## Imports

```python
import sqlite3
import pandas as pd

# Connect to today's practice database
conn = sqlite3.connect('dataset.db')

# Helper: run a query and show it as a clean table
def run(query):
    return pd.read_sql_query(query, conn)
```

## Theory
SQL processes a query in this logical order, not top-to-bottom as written:

1. `FROM` — pick the table
2. `WHERE` — filter the rows
3. `SELECT` — choose the columns
4. `ORDER BY` — sort what's left

`AND` requires every condition to be true (stricter, fewer rows). `OR` requires only one condition to be true (looser, more rows). `NULL` means *unknown* — it can only be tested with `IS NULL` / `IS NOT NULL`, never `=`.

## Example 1 — Filtering with WHERE and AND
Find right-footed Left Wingers.

```python
run("""
SELECT name, primary_position, preferred_foot
FROM player_stats
WHERE primary_position = 'Left Winger'
  AND preferred_foot = 'Right';
""")
```

## Example 2 — Sorting with ORDER BY
Right-footed players, most goals first.

```python
run("""
SELECT name, goals
FROM player_stats
WHERE preferred_foot = 'Right'
ORDER BY goals DESC;
""")
```

## Practice Exercise
Write a query that retrieves `name`, `assists`, and `primary_position` from `player_stats`, filtered to players with more than 4 assists **and** who are right-footed.

```python
# Your attempt:
run("""
SELECT name, assists, primary_position
FROM player_stats
WHERE assists > 4
  AND preferred_foot = 'Right';
""")
```

## Mini Challenge
Write a query that returns the **unique** `primary_position` values for players who scored 0 goals **or** have a missing (`NULL`) preferred foot.

```python
# Your attempt:
run("""
SELECT DISTINCT primary_position
FROM player_stats
WHERE goals = 0
   OR preferred_foot IS NULL;
""")
```

## Summary
- `WHERE` filters rows; `ORDER BY` sorts them — always in that order in the query.
- `AND` = fewer/stricter results, `OR` = more/looser results.
- `NULL` needs `IS NULL` / `IS NOT NULL` — `=` will silently miss it.
- `DISTINCT` is a quick way to see what unique values exist in a column.
